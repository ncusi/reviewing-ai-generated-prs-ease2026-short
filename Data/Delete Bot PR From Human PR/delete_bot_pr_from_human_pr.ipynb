```python
import pandas as pd

# Load parquet
input_path = "../human_pull_request.parquet"
df = pd.read_parquet(input_path)
```

```python
# Conditions
bots = df["user"].str.contains(r"\[bot\]", regex=True, na=False)

keywords = (
    df["body"].str.contains("PR-Codex", case=False, na=False) |
    df["body"].str.contains("Copilot", case=False, na=False) |
    df["body"].str.contains("Claude", case=False, na=False) |
    df["body"].str.contains("Devin", case=False, na=False) |
    df["body"].str.contains("Cursor", case=False, na=False)
)
```

```python
# SQL Condition: where NOT user LIKE '%[bot]%' AND (body like ...) OR user LIKE '%[bot]%'
condition = (~bots & keywords) | bots
```

```python
# Choose id column after filtering
ids_to_remove = df.loc[condition, "id"]

print(ids_to_remove)
```

```python
# Delete filtered PRs
df_filtered = df[~df["id"].isin(ids_to_remove)]
```

```python
# Write to new parquet
output_path = input_path.replace(".parquet", "_removed_bot_prs.parquet")
df_filtered.to_parquet(output_path, index=False)

print(f"Writing filtered file: {output_path}")
```
